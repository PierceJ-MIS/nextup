```python
def popular_games_by_genre(games_df, genre, n=5):
    # Identify the correct column for total reviews
    review_col = [col for col in games_df.columns if "review" in col.lower() and "total" in col.lower()]
    review_col = review_col[0] if review_col else "reviews_total"

    # Filter the games where the genre column contains the requested genre
    filtered_games = games_df[games_df["genres"].str.contains(genre, case=False, na=False)]
    # Sort the games by reviews to find the most popular ones
    sorted_games = filtered_games.sort_values(review_col, ascending=False)
    # Return the name, genres, and total reviews for the top n games
    return sorted_games[["name", "genres", review_col]].head(n)

my_function = popular_games_by_genre
my_result = popular_games_by_genre(games, "Action", 5)
print(my_result)
```
